Q1 – Présentation de l’algorithme : Régression linéaire
La régression linéaire (Linear Regression) est un algorithme d’apprentissage supervisé appartenant à la famille des méthodes de régression.
Son objectif est de prédire une valeur numérique continue à partir d’une ou plusieurs variables explicatives.
Elle repose sur l’hypothèse qu’il existe une relation linéaire entre les variables indépendantes et la variable à prédire.
L’algorithme cherche à trouver la droite ou l’hyperplan qui représente le mieux les données.
Il minimise la somme des carrés des écarts entre les valeurs réelles et les valeurs prédites.
Sa formule générale est : \(\hat{y}=\beta_0+\beta_1x_1+\cdots+\beta_px_p\).
Dans cette formule, \(\hat{y}\) représente la valeur prédite, \(\beta_0\) la constante, \(\beta_i\) les coefficients et \(x_i\) les variables explicatives.
Les coefficients sont les paramètres appris pendant l'entraînement du modèle.
Avantages : simplicité d'utilisation, rapidité d'exécution et facilité d'interprétation des résultats.
Limites : elle suppose une relation linéaire entre les variables et peut être sensible aux valeurs aberrantes.

```python
#Q2
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

fig, ax = plt.subplots(figsize=(10, 3))
ax.set_xlim(0, 10)
ax.set_ylim(0, 3)
ax.axis("off")

etapes = [
    (0.3, "Variables d'entrée\nX"),
    (3.7, "Apprentissage du modèle\nEstimation des coefficients"),
    (7.1, "Prédiction\nValeur de Y")
]

for x, texte in etapes:
    boite = FancyBboxPatch(
        (x, 1), 2.5, 1,
        boxstyle="round,pad=0.15",
        facecolor="lightblue",
        edgecolor="navy"
    )
    ax.add_patch(boite)
    ax.text(x + 1.25, 1.5, texte, ha="center", va="center")

ax.annotate("", xy=(3.6, 1.5), xytext=(2.9, 1.5),
            arrowprops=dict(arrowstyle="->", lw=2))
ax.annotate("", xy=(7, 1.5), xytext=(6.3, 1.5),
            arrowprops=dict(arrowstyle="->", lw=2))

plt.title("Fonctionnement de la régression linéaire")
plt.show()
```

Q1 – Présentation de l’algorithme : Ridge (Régression Ridge)
La régression Ridge (Ridge Regression) est un algorithme d’apprentissage supervisé appartenant à la famille des méthodes de régression régularisée.
Son objectif est de prédire une valeur numérique continue tout en limitant le surapprentissage (overfitting).
Elle est basée sur la régression linéaire, mais ajoute une pénalité qui limite la taille des coefficients du modèle.
Cette pénalité permet de réduire l'influence excessive de certaines variables explicatives.
L'algorithme cherche à minimiser les erreurs de prédiction tout en pénalisant les coefficients trop élevés.
Sa fonction de coût est : \(J(\beta)=\sum_{i=1}^{n}(y_i-\hat{y}_i)^2+\alpha\sum_{j=1}^{p}\beta_j^2\).
Dans cette formule, \(y_i\) représente la valeur réelle, \(\hat{y}_i\) la valeur prédite, \(\beta_j\) les coefficients et \(\alpha\) le paramètre de régularisation.
Plus alpha est élevé, plus les coefficients sont réduits.
Avantages : elle limite le surapprentissage et fonctionne bien lorsque les variables explicatives sont fortement corrélées.
Limites : elle ne supprime pas les variables inutiles et le choix de alpha influence les performances du modèle.

Q1 – Présentation de l’algorithme : Lasso (Régression Lasso)
La régression Lasso (Least Absolute Shrinkage and Selection Operator) est un algorithme d’apprentissage supervisé appartenant à la famille des méthodes de régression régularisée.
Son objectif est de prédire une valeur numérique continue tout en réduisant le nombre de variables explicatives utilisées par le modèle.
Elle est basée sur la régression linéaire, mais ajoute une pénalité qui limite la taille des coefficients.
Contrairement à Ridge, Lasso utilise une pénalité basée sur la valeur absolue des coefficients.
Cette méthode peut réduire certains coefficients jusqu'à zéro, ce qui permet d'effectuer une sélection automatique des variables.
Sa fonction de coût est : \(J(\beta)=\sum_{i=1}^{n}(y_i-\hat{y}_i)^2+\alpha\sum_{j=1}^{p}|\beta_j|\).
Dans cette formule, \(y_i\) représente la valeur réelle, \(\hat{y}_i\) la valeur prédite, \(\beta_j\) les coefficients et \(\alpha\) le paramètre de régularisation.
Plus alpha est élevé, plus la pénalisation est importante et davantage de coefficients peuvent devenir nuls.
Avantages : elle permet la sélection automatique des variables et limite le surapprentissage.
Limites : elle peut supprimer des variables qui contiennent des informations utiles et devient moins stable lorsque plusieurs variables sont fortement corrélées.

Contrairement à la régression linéaire simple, elle permet de représenter des relations sous forme de courbes.
Le modèle cherche à minimiser les écarts entre les valeurs réelles et les valeurs prédites.
Sa formule générale est : \(\hat{y}=\beta_0+\beta_1x+\beta_2x^2+\cdots+\beta_dx^d\).
Dans cette formule, \(\hat{y}\) représente la valeur prédite, \(x\) la variable explicative, \(\beta_i\) les coefficients et \(d\) le degré du polynôme.
Le degré détermine la complexité de la relation modélisée.
Avantages : elle permet de modéliser des relations non linéaires et d'obtenir des prédictions plus flexibles qu'une régression linéaire simple.
Limites : un degré trop élevé peut provoquer le surapprentissage et augmenter la complexité du modèle.

L’arbre de régression (Decision Tree Regressor) est un algorithme d’apprentissage automatique supervisé appartenant à la famille des méthodes de régression.

Son objectif est de prédire une valeur numérique continue à partir de plusieurs variables explicatives.

Son principe consiste à diviser progressivement les données en plusieurs groupes selon des conditions portant sur les variables explicatives.

Chaque division représente un nœud de décision, tandis que les branches représentent les différentes conditions.

L'algorithme choisit les divisions qui permettent de réduire au maximum les erreurs de prédiction à l'intérieur de chaque groupe.

Le processus continue jusqu'à atteindre les feuilles terminales, qui fournissent les valeurs prédites, généralement calculées à partir de la moyenne des observations de chaque feuille.

Le critère de division repose généralement sur la minimisation de l'erreur quadratique moyenne (MSE) :
\[
MSE=\frac{1}{n}\sum_{i=1}^{n}(y_i-\bar{y})^2
\]

Dans cette formule, \(n\) représente le nombre d'observations, \(y_i\) la valeur réelle de chaque observation et \(\bar{y}\) la moyenne des valeurs dans le groupe.

Les principaux hyperparamètres sont max_depth, qui définit la profondeur maximale de l'arbre, min_samples_split, qui détermine le nombre minimal d'observations nécessaires pour diviser un nœud, et min_samples_leaf, qui fixe le nombre minimal d'observations dans une feuille.

Avantages :
- Facile à comprendre et à interpréter grâce à sa structure arborescente.
- Capable de détecter des relations non linéaires entre les variables.
Limites :
- Sensible aux variations des données d'entraînement.
- Peut provoquer un surapprentissage lorsque l'arbre est trop profond.

La forêt aléatoire en régression (Random Forest Regressor) est un algorithme d’apprentissage automatique supervisé appartenant à la famille des méthodes de régression par ensemble (Ensemble Learning).

Son objectif est de prédire une valeur numérique continue en combinant les résultats de plusieurs arbres de décision.

Son principe consiste à construire plusieurs arbres de régression à partir de différents échantillons aléatoires des données d'entraînement.

Chaque arbre réalise une prédiction indépendamment des autres arbres, en utilisant également un sous-ensemble aléatoire de variables explicatives lors des divisions.

La prédiction finale est obtenue en calculant la moyenne des prédictions de tous les arbres de la forêt.

Cette combinaison permet de réduire la variance et de limiter le risque de surapprentissage par rapport à un arbre de décision unique.

La prédiction finale peut être représentée par la formule suivante :
\[
\hat{y}=\frac{1}{B}\sum_{b=1}^{B}T_b(X)
\]

Dans cette formule, \(B\) représente le nombre total d'arbres, \(T_b(X)\) la prédiction du \(b\)-ième arbre et \(\hat{y}\) la valeur finale prédite.

Les principaux hyperparamètres sont n_estimators, qui définit le nombre d'arbres, max_depth, qui limite leur profondeur, et max_features, qui détermine le nombre de variables considérées lors de chaque division.

Avantages :
- Offre généralement de bonnes performances grâce à la combinaison de plusieurs arbres.
- Réduit le risque de surapprentissage et améliore la stabilité des prédictions par rapport à un arbre unique.
Limites :
- Nécessite davantage de temps et de ressources informatiques lorsqu'un grand nombre d'arbres est utilisé.
- Moins facile à interpréter qu'un arbre de décision individuel.

Le Gradient Boosting en régression (Gradient Boosting Regressor) est un algorithme d’apprentissage automatique supervisé appartenant à la famille des méthodes de régression par ensemble (Ensemble Learning).

Son objectif est de prédire une valeur numérique continue en combinant progressivement plusieurs modèles faibles, généralement des arbres de décision.

Son principe consiste à construire une succession d'arbres de décision, dans laquelle chaque nouvel arbre cherche à corriger les erreurs commises par les arbres précédents.

Contrairement à la forêt aléatoire, qui construit plusieurs arbres indépendamment, le Gradient Boosting construit les arbres de manière séquentielle, chaque modèle dépendant des résultats du précédent.

L'algorithme utilise le gradient de la fonction de perte pour déterminer la direction dans laquelle les erreurs doivent être corrigées afin d'améliorer progressivement les prédictions.

La fonction objective peut être représentée par :
\[
J(F)=\sum_{i=1}^{n}L(y_i,F(x_i))
\]

Dans cette formule, \(n\) représente le nombre d'observations, \(y_i\) la valeur réelle, \(F(x_i)\) la valeur prédite et \(L\) la fonction de perte qui mesure l'erreur de prédiction.

La mise à jour du modèle peut être exprimée par :
\[
F_m(x)=F_{m-1}(x)+\eta h_m(x)
\]

Où \(F_m(x)\) représente le nouveau modèle, \(F_{m-1}(x)\) le modèle précédent, \(h_m(x)\) le nouvel arbre et \(\eta\) le taux d'apprentissage (learning rate).

Les principaux hyperparamètres sont max_iter, qui définit le nombre d'itérations, learning_rate, qui contrôle la contribution de chaque nouvel arbre, et max_leaf_nodes, qui limite le nombre de feuilles par arbre.

Avantages :
- Offre généralement de bonnes performances en capturant les relations complexes et non linéaires entre les variables.
- Corrige progressivement les erreurs des modèles précédents, ce qui permet d'améliorer la précision des prédictions.
Limites :
- Peut être sensible au choix des hyperparamètres et nécessiter plusieurs essais pour obtenir de bonnes performances.
- L'entraînement séquentiel peut être coûteux en temps de calcul et le modèle est difficile à interpréter.

La régression logistique est un algorithme d’apprentissage automatique supervisé utilisé pour résoudre des problèmes de classification. Contrairement à la régression linéaire, qui prédit une valeur numérique continue, la régression logistique permet de prédire une classe ou une catégorie.

Elle est principalement utilisée pour la classification binaire, c’est-à-dire lorsque la variable cible possède deux catégories (0 ou 1). Elle peut également être adaptée aux problèmes de classification multiclasse.

Son principe consiste à calculer une combinaison linéaire des variables explicatives, puis à appliquer la fonction sigmoïde afin d’obtenir une probabilité comprise entre 0 et 1.

La fonction sigmoïde est définie par :
\[
P(Y=1|X)=\frac{1}{1+e^{-z}}
\]

Avec :
\[
z=\beta_0+\beta_1X_1+\beta_2X_2+\cdots+\beta_nX_n
\]

- \(P(Y=1|X)\) : probabilité que l'observation appartienne à la classe 1.
- \(\beta_0\) : constante du modèle.
- \(\beta_i\) : coefficients associés aux variables explicatives.
- \(X_i\) : variables explicatives.
- \(e\) : base du logarithme népérien.
La classification est ensuite effectuée à partir d'un seuil, généralement fixé à 0,5 : si la probabilité est supérieure ou égale à 0,5, l'observation est affectée à la classe 1 ; sinon, elle est affectée à la classe 0.

Avantages :
- Simple à comprendre et à interpréter.
- Rapide à entraîner sur des jeux de données de taille modérée.
- Fournit des probabilités d'appartenance aux différentes classes.
- Adaptée aux problèmes de classification binaire et multiclasse.
Limites :
- Suppose une relation linéaire entre les variables explicatives et le logarithme des chances (log-odds).
- Peut être moins performante lorsque les relations entre les variables sont fortement non linéaires.
- Sensible aux valeurs aberrantes et à la multicolinéarité.
- Nécessite généralement une standardisation des variables lorsque leurs échelles sont différentes.

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```

import gymnasium as gym
from gymnasium import spaces

class OneStateBandit(gym.Env):
    metadata = {"render_modes": []}
    def __init__(self, probs=(.15,.35,.60)):
        super().__init__()
        self.probs = np.asarray(probs, dtype=float)
        self.action_space = spaces.Discrete(len(self.probs))
        self.observation_space = spaces.Discrete(1)
    def reset(self, seed=None, options=None):
        super().reset(seed=seed)
        return 0, {}
    def step(self, action):
        reward = float(self.np_random.random() < self.probs[int(action)])
        return 0, reward, True, False, {}

def run_epsilon_greedy(env, episodes=5000, epsilon=.10, seed=GRAINE):
    rng = np.random.default_rng(seed)
    q = np.zeros(env.action_space.n)
    counts = np.zeros(env.action_space.n, dtype=int)
    rewards = []
    for ep in range(episodes):
        state, _ = env.reset(seed=seed+ep)
        if rng.random() < epsilon:
            action = int(rng.integers(env.action_space.n))
        else:
            action = int(np.argmax(q))
        _, reward, terminated, truncated, _ = env.step(action)
        counts[action] += 1
        q[action] += (reward - q[action]) / counts[action]
        rewards.append(reward)
    return q, counts, np.asarray(rewards)

import gymnasium as gym

def entrainer_q_learning(episodes=8000, seed=GRAINE):
    env = gym.make("FrozenLake-v1", is_slippery=False)
    Q = np.zeros((env.observation_space.n, env.action_space.n))
    rng = np.random.default_rng(seed)
    alpha, gamma = .15, .95
    for ep in range(episodes):
        state, _ = env.reset(seed=seed+ep)
        epsilon = max(.03, 1 - ep/(episodes*.75))
        for _ in range(100):
            if rng.random() < epsilon:
                action = int(rng.integers(env.action_space.n))
            else:
                action = int(np.argmax(Q[state]))
            next_state, reward, terminated, truncated, _ = env.step(action)
            target = reward if terminated else reward + gamma*np.max(Q[next_state])
            Q[state, action] += alpha*(target-Q[state, action])
            state = next_state
            if terminated or truncated:
                break
    env.close()
    return Q

def entrainer_sarsa(episodes=8000, seed=GRAINE):
    env = gym.make("FrozenLake-v1", is_slippery=False)
    Q = np.zeros((env.observation_space.n, env.action_space.n))
    rng = np.random.default_rng(seed)
    alpha, gamma = .15, .95
    def choisir_action(s, epsilon):
        return int(rng.integers(env.action_space.n)) if rng.random() < epsilon else int(np.argmax(Q[s]))
    for ep in range(episodes):
        state, _ = env.reset(seed=seed+ep)
        epsilon = max(.03, 1 - ep/(episodes*.75))
        action = choisir_action(state, epsilon)
        for _ in range(100):
            next_state, reward, terminated, truncated, _ = env.step(action)
            next_action = choisir_action(next_state, epsilon) if not (terminated or truncated) else 0
            target = reward if terminated else reward + gamma*Q[next_state, next_action]
            Q[state, action] += alpha*(target-Q[state, action])
            state, action = next_state, next_action
            if terminated or truncated:
                break
    env.close()
    return Q

Q_ql = entrainer_q_learning()
Q_sarsa = entrainer_sarsa()
print("Politique Q-learning :", Q_ql.argmax(axis=1))
print("Politique SARSA :", Q_sarsa.argmax(axis=1))

def evaluer_politique(Q, episodes=500, seed=1234):
    env = gym.make("FrozenLake-v1", is_slippery=False)
    succes = []
    longueurs = []
    for ep in range(episodes):
        state, _ = env.reset(seed=seed+ep)
        total = 0
        for step in range(100):
            action = int(np.argmax(Q[state]))
            state, reward, terminated, truncated, _ = env.step(action)
            total += reward
            if terminated or truncated:
                break
        succes.append(total > 0)
        longueurs.append(step+1)
    env.close()
    return np.mean(succes), np.mean(longueurs)

Discussion à rédiger à partir des valeurs réellement obtenues
1. Régression : comparer MAE, RMSE et R². La régression linéaire est une référence simple ; la forêt et le Gradient Boosting capturent des relations non linéaires mais sont moins interprétables. Les résultats dépendent de la période de test et du caractère temporel des données.
2. Classification : comparer F1, précision, rappel, ROC-AUC et PR-AUC. Une forte précision peut se faire au prix d'un rappel plus faible. En gestion du risque de crédit, le seuil de décision doit tenir compte du coût des faux négatifs et faux positifs ; la métrique seule ne suffit pas.
3. Clustering : comparer silhouette, part de bruit et profils RFM. La silhouette ne valide pas la pertinence métier. Examiner la taille, la stabilité et le sens économique de chaque groupe.
4. Réduction de dimension : l'ACP permet de quantifier la variance expliquée ; t-SNE et UMAP servent surtout à préserver les voisinages locaux. Ne pas interpréter leurs axes comme des facteurs économiques.
5. Anomalies : sans étiquettes fiables, commenter la proportion d'alertes et leur profil, sans annoncer de précision ou rappel.
6. Renforcement : comparer le taux de réussite et la récompense cumulée sur plusieurs graines. Une seule graine peut donner une image trompeuse. DQN est facultatif ici ; le bandit custom est une simulation pédagogique et ne remplace pas un benchmark Gymnasium public.
7. Coût et interprétabilité : les modèles linéaires sont généralement rapides et lisibles ; KNN peut être coûteux en prédiction ; les ensembles d'arbres peuvent être performants mais moins explicables ; t-SNE/UMAP et les méthodes RL demandent des choix expérimentaux supplémentaires.
Limites méthodologiques à déclarer
- Les jeux UCI ne sont pas des données marocaines ; les conclusions ne doivent pas être généralisées directement au Maroc.
- Le jeu de crédit porte sur des clients taïwanais en 2005 ; il ne représente pas les pratiques actuelles d'une banque.
- Online Retail correspond à une entreprise britannique sur 2010–2011 ; les segments RFM sont propres à ce contexte historique.
- Wine est petit (178 observations) ; les projections sont exploratoires.
- Appliances Energy Prediction est une série de mesures sur un seul logement pendant environ 4,5 mois ; la généralisation à d'autres bâtiments est limitée.
- Les résultats des algorithmes ne doivent pas être présentés comme des résultats économiques réels ni comme des décisions prêtes à être déployées.
